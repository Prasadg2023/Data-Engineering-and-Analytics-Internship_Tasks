## Q1. Find the Frequency of Each Word in a Sentence
### Write a Python program to count how many times each word appears in a given sentence and store the result in a dictionary.

Start

Take a sentence as input

Convert the sentence into lowercase

Split the sentence into words

Count how many times each word appears

Store each word and its count in a dictionary

Return the dictionary

Stop

```python
def freq_cnt(a):
    words = a.lower().split()
    cnt = {x: words.count(x) for x in words}
    return cnt
```

```python
x = freq_cnt("python is easy and python is powerful")
### Additional Test case
z = freq_cnt("Python is interesting and python is easy")
print(x)
print(z)
```

## Q3. Find the First Non-Repeating Character in a Word
### Write a Python program to find the first character that appears only once in a given word. If no such character exists, return None.




Start

Take a word as input

Check each character in the word

Count how many times the character appears

If the count is one return that character

If no character appears only once return None

Stop

```python
def first_non_repeating(a):
    for x in a:
        if a.count(x) == 1:
            return x
    return None
```

```python
x = first_non_repeating("swiss")
### Additional Test case
z = first_non_repeating("aabb")
print(x)
print(z)
```

## Q6. Move All Zeros to the End of a List
### Write a Python program to move all zeros to the end of a list while maintaining the original order of non-zero elements.




Start

Take a list as input

Create an empty list for non-zero numbers

Create an empty list for zeros

Check each number in the list

If the number is zero add it to the zero list

Otherwise add it to the non-zero list

Join both lists

Return the final list

Stop

```python
def move_zeros(lst):
    non_zero_lst = []
    zero_lst = []
    for num in lst:
        if num == 0:
            zero_lst.append(num)
        else:
            non_zero_lst.append(num)
    return non_zero_lst + zero_lst
            
```

```python
x = move_zeros([0, 3, 0, 1, 2, 0, 4])
### Additional Test case
z = move_zeros([1, 2, 3, 0, 4, 0, 5])
print(x)
print(z)
```

## Q7. Find the Second Highest Distinct Number in a List
### Write a Python program to find the second-largest distinct number in a given list.
### Duplicate values should not be considered separate ranks.
### If a second distinct number does not exist, return None.



Start

Take a list of numbers as input

Remove duplicate numbers

Check if the list has less than two distinct numbers

If yes return None

Sort the numbers in descending order

Return the second number

Stop

```python
def second_largest(a):
    unique_numbers = list(set(a))
    if len(unique_numbers) < 2:
        return None
    unique_numbers.sort(reverse=True)
    return unique_numbers[1]
```

```python
x = second_largest([10, 20, 40, 30, 40, 20])
### Additional Test case
z = second_largest([5, 5, 5, 4, 3])
print(x)
print(z)
```

## Q8. Remove Duplicates Without Changing the Original Order
### Write a Python program to remove duplicate elements from a list while preserving the order of their first occurrence.



Start

Take a list as input

Create an empty list

Check each element in the original list

If the element is not in the new list add it

Return the new list

Stop

```python
def after_removal(lst):
    dup = []
    for i in lst:
        if i not in dup:
            dup.append(i)
    return dup
```

## Q9. Find the Missing Number from 1 to N
### Write a Python program to find the missing number in a list containing numbers from 1 to N.
### The list contains exactly one missing number, and all remaining numbers appear only once.



Start

Take a list and N as input

Calculate the sum of numbers from 1 to N

Calculate the sum of the given list

Subtract the list sum from the total sum

Return the missing number

Stop

```python
def missing_number(nums, n):
    total = n * (n + 1) // 2
    return total - sum(nums)
```

```python
x = missing_number([1, 2, 4, 5, 6], 6)
### Additional Test case
z = missing_number([1, 3, 4, 5], 5)
print(x)
print(z)
```

## Q10. Find Two Numbers Whose Sum Equals a Target
### Write a Python program to find the first pair of numbers whose sum equals a given target.
### Return the pair as a list. If no valid pair exists, return an empty list.
### Return the pair whose second element is encountered earliest when scanning the input from left to right. If multiple earlier elements qualify, choose the earliest one.



Start

Take a list and target as input

Check each number in the list

Compare it with the numbers after it

If the sum equals the target return both numbers as a list

If no pair is found return an empty list

Stop

```python
def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [nums[i], nums[j]]
    return []
```

```python
x = two_sum([2, 7, 11, 15], 9)
### Additional Test case
z = two_sum([1, 2, 3, 4, 5], 8)
print(x)
print(z)
```

## Q11. Find the Number Closest to Zero
### Write a Python program to find the number closest to zero in a given list.
### If two numbers are equally close to zero, return the positive number.


Start

Take a list of numbers as input

Find the absolute value of each number

Compare the distances from zero

If two numbers have the same distance choose the positive number

Return the number closest to zero

Stop

```python
def closest_to_zero(nums):
    return min(nums, key=lambda x: (abs(x), -x))
```

```python
x = closest_to_zero([-8, -3, 2, 5, -2])
### Additional Test case
z = closest_to_zero([-5, -1, 3, 2, 1])
print(x)
print(z)
```

## Q13. Rotate a List to the Right by K Positions
### Write a Python program to rotate the elements of a list to the right by K positions.
### Elements shifted beyond the last position should move to the beginning of the list.


Start

Take a list and K as input

Find the remainder of K divided by the length of the list

Take the last K elements

Take the remaining elements

Join the last elements at the beginning

Return the rotated list

Stop

```python
def rotate_right(nums, k):
    k = k % len(nums)
    return nums[-k:] + nums[:-k] if k else nums
```

```python
x = rotate_right([1, 2, 3, 4, 5], 2)
### Additional Test case
z = rotate_right([1, 2, 3, 4, 5], 7)
print(x)
print(z)
```

## Q14. Find the Maximum Sum of Two Consecutive Elements
### Write a Python program to find the maximum sum of any two consecutive elements in a list.
### Assume the list contains at least two elements.


```python
def max_consecutive_sum(nums):
    max_sum = nums[0] + nums[1]

    for i in range(1, len(nums) - 1):
        current_sum = nums[i] + nums[i + 1]
        if current_sum > max_sum:
            max_sum = current_sum

    return max_sum
```

```python
x = max_consecutive_sum([4, 7, 2, 9, 5])    
### Additional Test case
z = max_consecutive_sum([10, 20, 30, 40, 50])
print(x)
print(z)
```

## Q15. Find All Pairs of Numbers Whose Sum Equals a Target
### Write a Python program to find all unique pairs of numbers whose sum equals a given target.
### Each pair should contain two distinct list positions, and duplicate pairs should not be included.
### Return each pair in ascending numerical order. Return the list of pairs sorted by their first element. If no valid pairs exist, return an empty list.



Start

Take a list and target as input

Create an empty list to store pairs

Check each number with the numbers after it

If the sum equals the target create a pair

Sort the pair in ascending order

Add the pair if it is not already in the result

Sort all pairs by their first element

Return the list of pairs

Stop

```python
def all_pairs(nums, target):
    pairs = []
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])
                if pair not in pairs:
                    pairs.append(pair)

    pairs.sort()
    return pairs
```

```python
x = all_pairs([2, 4, 3, 5, 7, 8, 1], 9)
### Additional Test case
z = all_pairs([1, 2, 3, 4, 5], 6)
print(x)
print(z)
```
