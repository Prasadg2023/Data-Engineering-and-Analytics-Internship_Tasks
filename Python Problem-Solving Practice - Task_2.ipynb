## Q1. Find the Shortest Word in a Sentence
### Write a Python program to find the shortest word in a given sentence.


### 1. Observe

Input is a sentence

Output is the shortest word in the sentence

### 2. Understand

We have to find the word with the smallest length

### 3. Break

- Store the sentence in a variable
- Split the sentence into words
- Take the first word as the shortest word
- Check each word one by one
- If a word is shorter than the current shortest word then update it
- Print the shortest word

### 4. Logic

First we split the sentence into words using `split()`

Then we store the first word in the `short` variable

We use a `for` loop to check each word

If the length of a word is less than the length of `short` then we update `short`

At the end we print the shortest word

```python
sent = "Python makes coding easy"
words = sent.split()
short = words[0]
for word in words:
    if len(word) < len(short):
        short = word
print(short) 
```
